## Розділ B — Введення та перетворення типів

### Завдання 3 — Celsius to Fahrenheit

Запитайте у користувача температуру в **Цельсіях** (перетворіть введення на `float`). Використайте формулу:

$$F = C \times \frac{9}{5} + 32$$

Виведіть результат, округлений до **1 знака після коми**, через f-string. Приклад:

```
Temperature in Celsius: 21
21.0°C is 69.8°F
```

**Підказка:** щоб округлити до 1 знака після коми у f-string, використайте `{fahrenheit:.1f}`.

````python
# Запитуємо у користувача температуру в Цельсіях та конвертуємо у float
celsius = float(input("Temperature in Celsius: "))

# Обчислюємо температуру за Фаренгейтом за формулою (Celsius * 9/5) + 32
fahrenheit = (celsius * 9 / 5) + 32

# Виводимо результат з одним знаком після коми через f-string
print(f"{celsius:.1f}°C is {fahrenheit:.1f}°F")
````

## Розділ C — Арифметика з `//`, `%` та `**`

### Завдання 4 — Pizza party math

Запитайте у користувача два цілі числа (`int`):

- `totalSlices` — скільки всього шматочків піци у вас є
- `numFriends` — скільки друзів ділять піцу

Використайте `//`, щоб обчислити, скільки шматочків отримує кожен друг, і `%`, щоб дізнатися, скільки шматочків залишиться. Виведіть обидва значення через f-string. Приклад:

```
Total slices: 22
Number of friends: 4
Each friend gets 5 slices.
Leftover slices: 2
```

````python
# Запитуємо загальну кількість шматочків та кількість друзів (тип int)
totalSlices = int(input("Total slices: "))
numFriends = int(input("Number of friends: "))

# Обчислюємо кількість шматочків на кожного (//) та залишок (%)
slicesEach = totalSlices // numFriends
leftover = totalSlices % numFriends

# Виводимо результати через f-strings англійською
print(f"Each friend gets {slicesEach} slices.")
print(f"Leftover slices: {leftover}")
````

### Завдання 5 — Power calculator

Запитайте у користувача `base` (int) та `exponent` (int). Використайте `**`, щоб піднести `base` до степеня `exponent`. Виведіть результат через f-string.

Приклад:

```
Base: 2
Exponent: 10
2 to the power of 10 is 1024.
```

````python
# Запитуємо у користувача основу та показник степеня (обидва типу int)
base = int(input("Base: "))
exponent = int(input("Exponent: "))

# Обчислюємо результат, підносячи base до степеня exponent за допомогою **
result = base ** exponent

# Виводимо результат за допомогою f-string англійською мовою
print(f"{base} to the power of {exponent} is {result}")
````

## Розділ D — f-strings та форматування чисел

### Завдання 6 — Discount calculator

Запитайте у користувача `originalPrice` (float) і `discountPercent` (float, наприклад `25` для 25%). Обчисліть:

- суму знижки в доларах
- кінцеву ціну зі знижкою

Виведіть обидва значення через f-string з **2 знаками після коми**. Приклад:

```
Original price: 80
Discount %: 25
You save $20.00
Sale price: $60.00
```

**Підказка:** щоб округлити до 2 знаків після коми у f-string, використайте `{value:.2f}`.

````python
# Запитуємо оригінальну ціну та відсоток знижки (обидва типу float)
originalPrice = float(input("Original price: "))
discountPercent = float(input("Discount %: "))

# Обчислюємо суму економії (savings) та кінцеву ціну (salePrice)
# Формула: (ціна * відсоток) / 100
savings = (originalPrice * discountPercent) / 100
salePrice = originalPrice - savings

# Виводимо результати з округленням до 2 знаків після коми
print(f"You save ${savings:.2f}")
print(f"Sale price: ${salePrice:.2f}")
````

### Завдання 7 — Fix the broken program

Програма нижче має запитати рік народження користувача і сказати, скільки років залишилося до того, як йому виповниться 100. У ній **щонайменше три** помилки. Скопіюйте її у порожню клітинку, виправте та запустіть свою версію.

```python
currentYear = 2026
firstName = input("What's your name? ")
birthYear = input("What year were you born? ")
age = currentYear - birthYear
yearsTo100 = 100 - age
print("Hi " + firstName + ", you have " + yearsTo100 + " years until you turn 100!")
```

Подумайте над цим:
- Що завжди повертає `input()`? (What does `input()` always return?)
- Чи можна використовувати `+`, щоб приклеїти `int` до `str`? (Can you use `+` to glue an `int` onto a string?)
- Чи може f-string зробити останній рядок чистішим? (Could an f-string make the last line cleaner?)

````python
# Виправлений код: додано перетворення типу int() та f-string
currentYear = 2026
firstName = input("What's your name? ")

# Помилка 1: input() повертає str, тому додаємо int()
birthYear = int(input("What year were you born? "))

# Обчислюємо вік
age = currentYear - birthYear

# Обчислюємо скільки років залишилося до 100
yearsTo100 = 100 - age

# Помилка 2 та 3: Використовуємо f-string замість оператора +
# Це дозволяє не перетворювати числа на str вручну
print(f"Hi {firstName}, you have {yearsTo100} years until you turn 100!")
````

### Завдання 8 — Mini receipt generator (capstone)

Напишіть невелику програму, яка запитує у користувача:

- `itemName` (str)
- `quantity` (int)
- `unitPrice` (float)
- `taxPercent` (float) — наприклад `7.25` для 7.25%

Потім обчисліть:

- `subtotal = quantity * unitPrice`
- `tax = subtotal * (taxPercent / 100)`
- `total = subtotal + tax`

Виведіть акуратний чек через f-strings, де всі суми у доларах округлені до **2 знаків після коми**. Приклад:

```
------------------------------
         RECEIPT
------------------------------
Item: Coffee
Quantity: 3
Unit price: $4.50
Subtotal: $13.50
Tax (7.25%): $0.98
Total: $14.48
------------------------------
```

Додайте на початку коду блок коментарів-заголовка з полями `Program:`, `Author:`, `Date:` та `Purpose:` (як в Уроці 01).

````python
# Program: Mini Receipt Generator
# Author: [Ваше Ім'я]
# Date: 01.05.2026
# Purpose: Prints a receipt with subtotal, tax, and total

# 1. Запитуємо дані у користувача (item name, quantity, unit price, tax percent)
itemName = input("Item name: ")
quantity = int(input("Quantity: "))
unitPrice = float(input("Unit price: "))
taxPercent = float(input("Tax %: "))

# 2. Обчислюємо проміжний підсумок, податок та загальну суму
subtotal = quantity * unitPrice
tax = subtotal * (taxPercent / 100)
total = subtotal + tax

# 3. Виводимо акуратний чек через f-strings
print("------------------------------")
print("         RECEIPT")
print("------------------------------")
print(f"Item: {itemName}")
print(f"Quantity: {quantity}")
print(f"Unit price: ${unitPrice:.2f}")
print(f"Subtotal: ${subtotal:.2f}")
print(f"Tax ({taxPercent}%): ${tax:.2f}")
print(f"Total: ${total:.2f}")
print("------------------------------")
````
